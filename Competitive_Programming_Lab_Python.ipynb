## Q04-EA-2D-RowMajor-DecBin
### Effective Address (Decimal + Binary) — 2D Row Major

```python
def to_binary(n):
    if n == 0:
        return "0"
    bits = []
    while n > 0:
        bits.append(str(n % 2))
        n //= 2
    return "".join(reversed(bits))

base, w, rows, r, c = map(int, input("Base, size, rows, row, col: ").split())
ea = base + ((r * rows) + c) * w
print(f"Decimal EA = {ea}")
print(f"Binary EA = {to_binary(ea)}")
```

## Q09-SeparateChaining

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

def count(root):
    if root is None:
        return 0
    return 1 + count(root.left) + count(root.right)

root = Node(1)
root.left = Node(2)
root.right = Node(3)
root.left.left = Node(4)
root.left.right = Node(5)

print(f"Node count = {count(root)}")
```

## Q13-Tree-LeafCount
### Binary Tree — Leaf Node Count

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

def leaf_count(root):
    if root is None:
        return 0
    if root.left is None and root.right is None:
        return 1
    return leaf_count(root.left) + leaf_count(root.right)

root = Node(1)
root.left = Node(2)
root.right = Node(3)
root.left.left = Node(4)
root.left.right = Node(5)

print(f"Leaf count = {leaf_count(root)}")
```

## Q14-Tree-InternalNodes
### Binary Tree — Internal Node Count

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

def internal(root):
    if root is None or (root.left is None and root.right is None):
        return 0
    return 1 + internal(root.left) + internal(root.right)

root = Node(1)
root.left = Node(2)
root.right = Node(3)
root.left.left = Node(4)
root.left.right = Node(5)

print(f"Internal nodes = {internal(root)}")
```

## Q15-Tree-Height
### Binary Tree — Height

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

def height(root):
    if root is None:
        return 0
    l = height(root.left)
    r = height(root.right)
    return 1 + max(l, r)

root = Node(1)
root.left = Node(2)
root.right = Node(3)
root.left.left = Node(4)

print(f"Height = {height(root)}")
```

## Q22-BFS
### Graph — Breadth First Search

```python
from collections import deque

adj = [[0] * 10 for _ in range(10)]
vis = [0] * 10
n = 0

def bfs(s):
    q = deque()
    vis[s] = 1
    q.append(s)
    order = []
    while q:
        u = q.popleft()
        order.append(u)
        for i in range(n):
            if adj[u][i] and not vis[i]:
                vis[i] = 1
                q.append(i)
    print("BFS: " + " ".join(map(str, order)))

n, e = map(int, input("Vertices, Edges: ").split())
for _ in range(e):
    u, v = map(int, input().split())
    adj[u][v] = adj[v][u] = 1

bfs(0)
```

## Q24-BellmanFord
### Bellman-Ford Shortest Path

```python
INF = 99999

n, e = map(int, input("Vertices, Edges: ").split())
eu, ev, ew = [], [], []
for _ in range(e):
    u, v, w = map(int, input().split())
    eu.append(u)
    ev.append(v)
    ew.append(w)

src = int(input("Source: "))

dist = [INF] * n
dist[src] = 0

for _ in range(n - 1):
    for j in range(e):
        if dist[eu[j]] != INF and dist[eu[j]] + ew[j] < dist[ev[j]]:
            dist[ev[j]] = dist[eu[j]] + ew[j]

for i in range(n):
    print(f"{src}->{i}: {dist[i]}")
```

## Q25-Dijkstra
### Dijkstra Shortest Path

```python
INF = 99999

n, e = map(int, input("Vertices, Edges: ").split())
adj = [[0] * 10 for _ in range(10)]
dist = [0] * 10
vis = [0] * 10

for _ in range(e):
    u, v, w = map(int, input().split())
    adj[u][v] = adj[v][u] = w

src = int(input("Source: "))

for i in range(n):
    dist[i] = INF
dist[src] = 0

for _ in range(n):
    mn, mu = INF, -1
    for j in range(n):
        if not vis[j] and dist[j] < mn:
            mn, mu = dist[j], j
    if mu == -1:
        break
    vis[mu] = 1
    for j in range(n):
        if adj[mu][j] and not vis[j] and dist[mu] + adj[mu][j] < dist[j]:
            dist[j] = dist[mu] + adj[mu][j]

for i in range(n):
    print(f"{src}->{i}: {dist[i]}")
```

## Q26-Prims
### Prim's Minimum Spanning Tree

```python
INF = 99999

n = int(input("Vertices: "))
print(f"Enter the {n}x{n} adjacency matrix:")
adj = [list(map(int, input().split())) for _ in range(n)]

key = [INF] * n
in_mst = [0] * n
parent = [-1] * n
key[0] = 0

for _ in range(n):
    mn, u = INF, -1
    for v in range(n):
        if not in_mst[v] and key[v] < mn:
            mn, u = key[v], v
    in_mst[u] = 1
    for v in range(n):
        if adj[u][v] and not in_mst[v] and adj[u][v] < key[v]:
            key[v] = adj[u][v]
            parent[v] = u

total = 0
for i in range(1, n):
    print(f"{parent[i]}-{i}: {adj[parent[i]][i]}")
    total += adj[parent[i]][i]

print(f"Total MST cost: {total}")
```

## Q27-Kruskal
### Kruskal's Minimum Spanning Tree

```python
parent = list(range(10))

def find(x):
    if parent[x] == x:
        return x
    return find(parent[x])

def unite(x, y):
    parent[find(x)] = find(y)

n, e = map(int, input("Vertices, Edges: ").split())
u = [0] * e
v = [0] * e
w = [0] * e
for i in range(e):
    u[i], v[i], w[i] = map(int, input().split())

for i in range(n):
    parent[i] = i

order = sorted(range(e), key=lambda i: w[i])

total = 0
for i in order:
    if find(u[i]) != find(v[i]):
        unite(u[i], v[i])
        print(f"{u[i]}-{v[i]}: {w[i]}")
        total += w[i]

print(f"Total MST cost: {total}")
```

## Q29-Tabulation
### 0/1 Knapsack — Dynamic Programming (Tabulation)

```python
n, W = map(int, input("Items, Capacity: ").split())

wt = [0] * (n + 1)
val = [0] * (n + 1)
for i in range(1, n + 1):
    wt[i], val[i] = map(int, input(f"Weight Value of item {i}: ").split())

dp = [[0] * (W + 1) for _ in range(n + 1)]
for i in range(n + 1):
    for j in range(W + 1):
        if i == 0 or j == 0:
            dp[i][j] = 0
        elif wt[i] <= j:
            dp[i][j] = max(val[i] + dp[i - 1][j - wt[i]], dp[i - 1][j])
        else:
            dp[i][j] = dp[i - 1][j]

print(f"Max value = {dp[n][W]}")
```

## Q35-PriceSorting-3Methods
### Price Sorting — Bubble, Selection and Insertion Sort

```python
def bubble(a):
    n = len(a)
    for i in range(n - 1):
        for j in range(n - i - 1):
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]

def selection(a):
    n = len(a)
    for i in range(n - 1):
        m = i
        for j in range(i + 1, n):
            if a[j] < a[m]:
                m = j
        a[i], a[m] = a[m], a[i]

def insertion(a):
    n = len(a)
    for i in range(1, n):
        k = a[i]
        j = i - 1
        while j >= 0 and a[j] > k:
            a[j + 1] = a[j]
            j -= 1
        a[j + 1] = k

n = int(input("N: "))
base = list(map(float, input().split()))
a = base.copy()
b = base.copy()
c = base.copy()

bubble(a)
print("Bubble: " + " ".join(f"{x:.2f}" for x in a))

selection(b)
print("Selection: " + " ".join(f"{x:.2f}" for x in b))

insertion(c)
print("Insertion: " + " ".join(f"{x:.2f}" for x in c))
```

## Q37-TernarySearch
### Ternary Search

```python
n = int(input("n: "))
a = list(map(int, input().split()))
key = int(input("Key: "))

lo, hi = 0, n - 1
found_at = -1
while lo <= hi:
    m1 = lo + (hi - lo) // 3
    m2 = hi - (hi - lo) // 3
    if a[m1] == key:
        found_at = m1
        break
    if a[m2] == key:
        found_at = m2
        break
    if key < a[m1]:
        hi = m1 - 1
    elif key > a[m2]:
        lo = m2 + 1
    else:
        lo = m1 + 1
        hi = m2 - 1

if found_at != -1:
    print(f"Found at {found_at}")
else:
    print("Not found")
```

## Q38-QuickSort
### Quick Sort

```python
def partition(a, l, r):
    p = a[r]
    i = l - 1
    for j in range(l, r):
        if a[j] <= p:
            i += 1
            a[i], a[j] = a[j], a[i]
    a[i + 1], a[r] = a[r], a[i + 1]
    return i + 1

def quick_sort(a, l, r):
    if l < r:
        pi = partition(a, l, r)
        quick_sort(a, l, pi - 1)
        quick_sort(a, pi + 1, r)

n = int(input("n: "))
a = list(map(int, input().split()))

quick_sort(a, 0, n - 1)
print(" ".join(map(str, a)))
```

## Q46-MissingID-XOR
### Find Missing ID using XOR

```python
n = int(input("N (total IDs from 1 to N): "))

xor_all = 0
for i in range(1, n + 1):
    xor_all ^= i

print(f"Enter {n - 1} recorded IDs: ")
xor_arr = 0
for _ in range(n - 1):
    x = int(input())
    xor_arr ^= x

print(f"Missing ID = {xor_all ^ xor_arr}")
```

## Q47-ScholarshipSelection
### Scholarship Selection

```python
n = int(input("N students: "))
att = [0.0] * n
gpa = [0.0] * n

print("Enter attendance and GPA for each student:")
for i in range(n):
    att[i], gpa[i] = map(float, input().split())

print("Scholarship eligible (attendance>=75 and GPA>=8.0):")
for i in range(n):
    if att[i] >= 75 and gpa[i] >= 8.0:
        print(f"Student {i + 1}")
```

## Q51-LibraryManagementSystem
### Library Management System

```python
class Book:
    def __init__(self, book_id, title):
        self.id = book_id
        self.title = title
        self.available = True

library = []

def add_book():
    book_id, title = input("ID Title: ").split()
    library.append(Book(int(book_id), title))
    print("Added")

def issue_book():
    book_id = int(input("Book ID: "))
    for book in library:
        if book.id == book_id:
            if book.available:
                book.available = False
                print("Issued")
            else:
                print("Not available")
            return
    print("Not found")

def return_book():
    book_id = int(input("Book ID: "))
    for book in library:
        if book.id == book_id:
            book.available = True
            print("Returned")
            return
    print("Not found")

def display():
    for book in library:
        status = "Available" if book.available else "Issued"
        print(f"{book.id} {book.title} {status}")

while True:
    ch = int(input("1.Add 2.Issue 3.Return 4.Display 0.Exit: "))
    if ch == 1:
        add_book()
    elif ch == 2:
        issue_book()
    elif ch == 3:
        return_book()
    elif ch == 4:
        display()
    if ch == 0:
        break
```

## Q52-CoinChange
### Coin Change — Greedy vs Dynamic Programming

```python
coins = [1, 5, 10, 25]
n = len(coins)

amount = int(input("Amount: "))

print("Greedy: ", end="")
rem = amount
for i in range(n - 1, -1, -1):
    while rem >= coins[i]:
        print(coins[i], end=" ")
        rem -= coins[i]
print()

dp = [0] * (amount + 1)
for i in range(1, amount + 1):
    dp[i] = 99999
    for j in range(n):
        if coins[j] <= i and dp[i - coins[j]] + 1 < dp[i]:
            dp[i] = dp[i - coins[j]] + 1

print(f"DP min coins: {dp[amount]}")
```
